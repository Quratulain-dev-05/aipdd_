# Real-Time Object Detection using YOLO
**AI Project Design and Development (AI-316) | Lab 01**

## Problem Statement
Manual monitoring of images and video feeds is **slow, expensive and error-prone**.
This project builds an AI-based pipeline that automatically **detects and classifies objects**
using the YOLO architecture, on top of a reproducible AI engineering environment.

## Core Objectives
- Set up an **isolated, reproducible** environment (`aipdd_env`) with pinned dependencies
- Maintain a **production-ready** repository structure under Git version control
- Collect and preprocess an image dataset
- Train and evaluate a YOLO object detection model

### Technology Stack
- **Language:** Python 3
- **Libraries:** NumPy, OpenCV, Matplotlib, Ultralytics (YOLO), PyTorch
- **Tools:** Jupyter Notebook, Git, venv

### Environment Setup
```bash
python -m venv aipdd_env
source aipdd_env/bin/activate        # Windows: aipdd_env\Scripts\activate
pip install -r requirements.txt
```
